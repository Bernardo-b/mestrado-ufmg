# Análise do Problema de Atribuição de Tarefas a Agentes

Este notebook tem como objetivo analisar o problema de otimização multiobjetivo de atribuição de tarefas a agentes, explorando os dados fornecidos e compreendendo a estrutura do problema.

## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.io import loadmat
import os

# Configurações para visualização
plt.style.use('default')
sns.set_palette("husl")
%matplotlib inline

print("Bibliotecas importadas com sucesso!")
```

## 2. Load and Explore Dataset Files

Vamos carregar todos os arquivos CSV e explorar sua estrutura:

```python
# Definir o caminho para os dados
data_path = "data/"

# Carregar cada arquivo CSV em uma variável específica
print("Carregando arquivos de dados...")
print("=" * 50)

# Matriz A - custos/tempos de execução (5 agentes x 50 tarefas)
A_matriz_recursos = pd.read_csv(os.path.join(data_path, "data_5x50_a.csv"), header=None)
print(f"Matriz A (recursos): {A_matriz_recursos.shape}")
print(f"  Primeiras linhas:")
print(A_matriz_recursos.head())
print()

# Vetor b - capacidades dos agentes (5 agentes)
b_vetor_capacidades = pd.read_csv(os.path.join(data_path, "data_5x50_b.csv"), header=None)
print(f"Vetor b (capacidades): {b_vetor_capacidades.shape}")
print(f"  Valores:")
print(b_vetor_capacidades.values.flatten())
print()

# Matriz C - custos alternativos/complexidade (5 agentes x 50 tarefas)
C_matriz_custos = pd.read_csv(os.path.join(data_path, "data_5x50_c.csv"), header=None)
print(f"Matriz C (custos alternativos): {C_matriz_custos.shape}")
print(f"  Primeiras linhas:")
print(C_matriz_custos.head())
print()

# Número de agentes (m)
m_agentes = pd.read_csv(os.path.join(data_path, "data_5x50_m.csv"), header=None).iloc[0, 0]
print(f"Número de agentes (m): {m_agentes}")

# Número de tarefas (n)
n_tarefas = pd.read_csv(os.path.join(data_path, "data_5x50_n.csv"), header=None).iloc[0, 0]
print(f"Número de tarefas (n): {n_tarefas}")

print("\n" + "=" * 50)
print("Resumo dos dados carregados:")
print(f"- {m_agentes} agentes × {n_tarefas} tarefas")
print(f"- Matriz A: {A_matriz_recursos.shape} (custos/tempos)")
print(f"- Vetor b: {b_vetor_capacidades.shape} (capacidades)")
print(f"- Matriz C: {C_matriz_custos.shape} (custos alternativos)")
```
